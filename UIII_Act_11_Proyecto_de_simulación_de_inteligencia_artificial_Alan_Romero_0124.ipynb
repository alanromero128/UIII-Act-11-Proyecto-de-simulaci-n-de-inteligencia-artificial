Diccionario de Arbitros

```python
diccionario_arbitros = {}

# 2. Función para simular la IA y evaluar al árbitro
def simular_ia(partidos, sanciones):
    # Lógica de IA basada en reglas (Sistemas Expertos)
    if partidos > 50 and sanciones == 0:
        return "RECOMENDADO: Torneo Internacional ⭐"
    elif partidos > 20 and sanciones <= 2:
        return "RECOMENDADO: Torneo Nacional 🇲🇽"
    elif partidos <= 20 and sanciones > 3:
        return "ALERTA: Requiere Capacitación ⚠️"
    else:
        return "RECOMENDADO: Torneo Local 🏟️"

# 3. Función para mostrar los datos en una tabla limpia y alineada
def mostrar_tabla():
    if not diccionario_arbitros:
        print("\n[!] La tabla está vacía. Registra árbitros primero.")
        return

    print("\n" + "=" * 115)
    # Encabezados de la tabla con alineación a la izquierda (<) y espacios fijos
    print(f"{'Deporte':<12} | {'Federación':<12} | {'Nacionalidad':<14} | {'Certificación':<15} | {'Partidos':<10} | {'Sanciones':<10} | {'Dictamen de IA'}")
    print("-" * 115)

    # Ciclo FOR para recorrer la lista de diccionarios
    for arbitro in diccionario_arbitros:
        # Ejecutamos la IA para cada árbitro antes de imprimir
        recomendacion = simular_ia(arbitro['partidos'], arbitro['sanciones'])

        print(f"{arbitro['deporte']:<12} | "
              f"{arbitro['federacion']:<12} | "
              f"{arbitro['nacionalidad']:<14} | "
              f"{arbitro['nivel']:<15} | "
              f"{arbitro['partidos']:<10} | "
              f"{arbitro['sanciones']:<10} | "
              f"{recomendacion}")
    print("=" * 115)

# 4. Menú Principal Interactivo (Ciclo While True)
while True:
    print("\n" + "•" * 15 + " SISTEMA DE GESTIÓN DE ÁRBITROS (IA) " + "•" * 15)
    print("1. ➕ Registrar Nuevo Árbitro")
    print("2. 📋 Ver Tabla de Rendimiento")
    print("3. ❌ Salir del Sistema")

    opcion = input("Selecciona una opción (1-3): ")

    if opcion == "1":
        print("\n--- CAPTURA DE DATOS DEL ÁRBITRO ---")

        # Entrada de datos (Key: Value)
        fede = input("Federación (ej. FIFA, FIBA): ")
        partidos_dirigidos = int(input("Cantidad de partidos dirigidos: "))
        certificacion = input("Nivel de certificación (ej. Local, Nacional, Elite): ")
        dep = input("Deporte: ")
        sanciones_recibidas = int(input("Sanciones o reportes aplicados: "))
        nac = input("Nacionalidad: ")


        arbitro_nuevo = {
            "federacion": fede,
            "partidos": partidos_dirigidos,
            "nivel": certificacion,
            "deporte": dep,
            "sanciones": sanciones_recibidas,
            "nacionalidad": nac
        }

        diccionario_arbitros.append(arbitro_nuevo)
        print("\n[✓] Árbitro registrado con éxito y analizado por la IA.")

    elif opcion == "2":
        mostrar_tabla()

    elif opcion == "3":
        print("\nCerrando el sistema de IA. ¡Hasta luego, Alan!")
        break

    else:
        print("\n[X] Opción no válida. Por favor, elige 1, 2 o 3.")
```
